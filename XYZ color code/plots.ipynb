```python
import pandas as pd 
import numpy as np
import matplotlib.pyplot as plt
df = pd.read_csv('T_mem.txt', sep=',', header= None, names=['x', 'y'],dtype= {'x': np.float64, 'y': np.float64})
T = df['x'].to_numpy()
N = df['y'].to_numpy()
print(T)
print(N)
```

```python
x1 =[ 12*9, 15*12, 18*15, 21*18, 24*21, 27*24, 30*27]
y1 = [27187, 36462, 70455, 80239, 188794, 457136, 407294]
```

```python
plt.figure(figsize=(10, 4))
plt.plot(T, N, 'o', label='reference')
plt.plot(x1, y1, 'o', label='sim', linewidth=2)
plt.xlabel('N')
plt.ylabel('T')
plt.xscale('log')
plt.yscale('log')
plt.title('T_mem vs N on error rate 1e-4')
plt.grid()
plt.legend()
plt.savefig('T_mem_vs_N.png')
plt.show()
```

```python
ratios = np.divide(N[:6], y1)
print('ratios:', ratios)
```

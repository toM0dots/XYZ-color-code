```python
import math
e = np.array(E)
l = 8
N = l*l
energy = np.arange(N)
error_rate = 0.01
beta = math.log((3+error_rate)/error_rate)/3
n, bins = np.histogram(e[8000::10], bins=np.arange(N+1))
centers = 0.5*(bins[:-1] + bins[1:])


dos = [math.comb(N, int(en)) for en in energy]
factor = dos* np.exp(-beta * energy)
P = factor/ sum(factor)
counts = len(E[8000::10])

eq_E = 1/ (1 + np.exp(beta)) * N
eq_E_MC = np.mean(E[8000::10])
```

```python
integral = 0
for i in range(64):
    integral += factor[i] * (bins[i+1] - bins[i])

print(integral)
```

```python
import numpy as np

with open('time_scale.txt', 'r') as f:
    text = f.read()

# turn the entire comma-separated blob into a 1D array of floats
data = np.fromstring(text, sep=',')
data1 = sorted(data)
plt.scatter(np.arange(len(data)), data1, color='blue', alpha=0.5, label='time scale')
plt.xlabel('Index')
plt.ylabel('Time Scale')
plt.title(r'Time Scale $\Delta t$')
plt.grid()
plt.xscale('log')
plt.yscale('log')
plt.legend()
plt.savefig('time_scale.png')
```
